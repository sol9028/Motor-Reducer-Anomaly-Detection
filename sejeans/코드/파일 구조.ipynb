```python
from pathlib import Path

def print_tree(dir_path: str | Path, prefix: str = ''):
    """지정된 디렉토리의 구조를 트리 형태로 출력하는 함수입니다."""
    dir_path = Path(dir_path)
    
    # 폴더가 존재하지 않으면 종료
    if not dir_path.exists() or not dir_path.is_dir():
        print(f"오류: '{dir_path}' 경로가 존재하지 않거나 폴더가 아닙니다.")
        return

    # 최상단 폴더 이름 출력 (최초 호출 시점)
    if prefix == '':
        print(f'📁 {dir_path.name}/')
    
    # 폴더 내의 항목(디렉토리 및 파일) 목록 가져오기 및 정렬 (폴더 우선)
    paths = list(dir_path.iterdir())
    paths.sort(key=lambda x: (x.is_file(), x.name))
    
    # 각 항목을 순회하며 구조 출력
    for i, path in enumerate(paths):
        is_last = (i == len(paths) - 1)
        
        # 현재 항목의 접두사 선 모양 결정
        connector = '└── ' if is_last else '├── '
        
        # 파일 또는 폴더 아이콘 설정
        icon = '📄 ' if path.is_file() else '📁 '
        
        print(f'{prefix}{connector}{icon}{path.name}')
        
        # 항목이 하위 폴더인 경우 재귀 호출로 내부 구조 탐색
        if path.is_dir():
            extension = '    ' if is_last else '│   '
            print_tree(path, prefix=prefix + extension)

# '샘플데이터' 폴더의 절대 경로 지정
# OS 정보에 따라 작업 중인 샘플데이터 폴더 경로로 설정되어 있습니다.
folder_path = r"c:\Users\kingm\OneDrive\문서\새 폴더\OneDrive\바탕 화면\학교\4학년 1학기\시스템분석\샘플데이터"

# 구조 출력 함수 실행
print_tree(folder_path)
```
